```python
import pandas as pd
data = {
    "StudyHours": [2, 4, None, 8, 10, None, 6],
    "Marks": [50, None, 70, 80, 90, 60, None],
    "Gender": ["Male", "Female", None, "Male", "Female", "Male", "Female"],
    "Result": ["Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Fail"]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
df["StudyHours"] = df["StudyHours"].fillna(
    df["StudyHours"].mean()
)
df["Marks"] = df["Marks"].fillna(
    df["Marks"].median()
)
df["Gender"] = df["Gender"].ffill()
df["Result"] = df["Result"].map({
    "Fail": 0,
    "Pass": 1
})
df = pd.get_dummies(
    df,
    columns=["Gender"],
    dtype=int
)

print("\nPreprocessed Dataset:")
print(df)
df.to_csv("students_preprocessed.csv", index=False)
```
