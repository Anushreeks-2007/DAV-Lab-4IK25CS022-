```python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# 1. EXPLORATORY DATA ANALYSIS
df = pd.read_csv("World Economic Indicators - Sheet1.csv")

print(df.head())
print(df.describe())
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())

# 2. DATA PREPROCESSING

# Mean Imputation
df = df.fillna(df.mean(numeric_only=True))

# Encoding Categorical Variables
df = pd.get_dummies(df, columns=["Region", "IncomeGroup"],
                    dtype=int)

# 3. OUTLIER DETECTION AND TREATMENT

col = "GDP"

# IQR Method
Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("IQR Outliers:")
print(df[(df[col] < lower) | (df[col] > upper)])

df[col] = df[col].clip(lower, upper)

# Z-Score Method
z = (df[col] - df[col].mean()) / df[col].std()

print("Z-Score Outliers:")
print(df[abs(z) > 3])

# 4. FEATURE ENGINEERING

features = [
    "Birth rate per 1000 people",
    "Death rate per 1000 people",
    "GDP",
    "GDP per capita"
]

X = df[features]

# Normalization
minmax = MinMaxScaler()
normalized = minmax.fit_transform(X)
print("Normalized Data:\n", normalized[:5])

# Standardization
scaler = StandardScaler()
standardized = scaler.fit_transform(X)
print("Standardized Data:\n", standardized[:5])

# PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(standardized)

print("PCA Results:\n", pca_result[:5])
print("Explained Variance:", pca.explained_variance_ratio_)
```
