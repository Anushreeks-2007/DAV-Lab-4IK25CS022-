```python
import pandas as pd
df=pd.read_csv("students.csv")
Q1=df["StudyHours"].quantile(0.25)   #4
Q3=df["StudyHours"].quantile(0.75)   #8
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
outliers_iqr = df[
    (df["StudyHours"]<lower)
    (df["StudyHours"]>upper)
    ]
print(f'IQR bounds: [{lower:.2f},{upper:.2f}]')
print('Outliers found (IQR):',len(outliers_iqr))
    
```
